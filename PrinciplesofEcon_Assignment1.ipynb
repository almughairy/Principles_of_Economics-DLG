```python
import numpy as np
import matplotlib.pyplot as plt

""" Principles of Economics """

# Price (per kilo)
price_per_kilo = np.array([1, 2, 3, 4, 5, 6, 7])
# Quantity supplied (per kilo)
quantity_supplied = np.array([55, 65, 75, 85, 95, 105, 115])
# Quantity demanded (per kilo)
quantity_demanded = np.array([130, 115, 100, 85, 70, 55, 40])

# Plotting standard economic supply and demand curves (Quantity on X, Price on Y)
plt.figure(figsize=(8, 6))
plt.plot(quantity_demanded, price_per_kilo, label='Demand Curve', color='blue', marker='x')
plt.plot(quantity_supplied, price_per_kilo, label='Supply Curve', color='orange', marker='x')

# Highlight the equilibrium point (Price = 4, Quantity = 85)
plt.plot(85, 4, 'ro', label='Equilibrium (85, $4)')
plt.axvline(x=85, color='gray', linestyle='--')
plt.axhline(y=4, color='gray', linestyle='--')

plt.xlabel('Quantity (kilos)')
plt.ylabel('Price (per kilo)')
plt.title("Market for Cinnamon Rolls")
plt.grid(True)
plt.legend()
plt.show()
```
