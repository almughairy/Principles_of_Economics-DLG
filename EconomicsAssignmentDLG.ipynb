```python
import xml.etree.ElementTree as ET

# The XML file is already unzipped in your directory.
filename = "/content/UNdata_Export_20261115_154039127.xml"

tree = ET.parse(filename)
root = tree.getroot()

# Collect Oman's values for 2004.
data = {}

for record in root.findall(".//record"):
    fields = {
        field.attrib["name"]: field.text
        for field in record.findall("field")
    }

    if fields.get("Country or Area") == "Oman" and fields.get("Year") == "2004":
        data[fields["Item"]] = float(fields["Value"].replace(",", ""))

gdp = data["Gross Domestic Product (GDP)"]

components = {
    "C": data[
        "Household consumption expenditure "
        "(including Non-profit institutions serving households)"
    ],
    "I": data["Gross capital formation"],
    "G": data["General government final consumption expenditure"],
    "X": data["Exports of goods and services"],
    "M": data["Imports of goods and services"],
}

components["NX"] = components["X"] - components["M"]

# Percentage of GDP = component / GDP × 100
for name, value in components.items():
    percentage = value / gdp * 100
    print(f"{name}: {percentage:.2f}%")

total = sum(components[name] for name in ["C", "I", "G", "NX"])
print(f"\nC + I + G + NX: {total / gdp * 100:.2f}%")
```
